## Errors

The next two cells raise errors on purpose. That is expected.

```python
a = 5
b = "a"
c = a b
```

Runtime error: adding an `int` and a `str` is not allowed (`TypeError`).


```python
a = 5
b = "a"
c = a + b
```

## Logic error (wrong intention)

Goal: calculate $5^2 = 25$. The next cell runs without an error, but the result is wrong: in Python `^` is bitwise XOR, not power.


Wrong operator (`^`): runs, but not what we meant.


```python
a = 5
b = 2
c = a ^ b
c
```

Correct operator (`**`):


```python
a = 5
b = 2
c = a ** b
c
```

## Floating-point error

Some decimals cannot be stored exactly. Equality checks can surprise you.


```python
result = 0.1 + 0.1 + 0.1 - 0.3
print(f"Result: {result:.20f}")
result == 0.0
```

In Python, `(1/3)*3` often equals `1.0` exactly. Compare with the cell above.


```python
result = (1/3) * 3
result == 1
```
